# Exploring the evolutionary relationship between CMD, AhpD, and DehI proteins

This notebook provides a step-by-step investigation of the relationships among three groups of structurally related proteins: the **carboxymuconolactone decarboxylase (CMD) family**, the **AhpD family**, and the **DehI-like 2-haloacid dehalogenases**. The central hypothesis is that the larger DehI fold arose through an ancient duplication and fusion of a smaller ancestral protein related to the CMD and/or AhpD groups. Under this model, the N- and C-terminal regions of DehI should retain detectable evidence of common ancestry with one another and with the smaller proteins.

The analysis develops this argument in stages. First, we inspect and quality-control the family-based sequence collections, using redundancy-reduced datasets so that densely sampled organisms do not dominate the comparisons. We then compare sequence lengths and domain organization to establish whether DehI has the architecture expected of a fusion product. Next, we examine the internal relationship between the two putative halves of DehI and compare each half independently with CMD and AhpD sequences. Finally, multiple-sequence alignments and phylogenetic reconstructions are used to ask whether the observed similarities are consistent with shared ancestry, duplication, and subsequent divergence.

Each step is intended to separate observation from interpretation. Structural resemblance or sequence similarity alone does not demonstrate a duplication-and-fusion event; support for the hypothesis should come from agreement among domain architecture, internal sequence symmetry, conserved residues, and phylogenetic placement. Alternative explanations—including convergence, remote homology without fusion, and biases introduced by sequence annotation or sampling—will therefore be considered alongside the proposed evolutionary model.
