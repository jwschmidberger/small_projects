# Exploring the evolutionary relationship between CMD, AhpD, and DehI proteins

This notebook provides a step-by-step investigation of the relationships among three groups of structurally related proteins: the **carboxymuconolactone decarboxylase (CMD) family**, the **AhpD family**, and the **DehI-like 2-haloacid dehalogenases**. The central hypothesis is that the larger DehI fold arose through an ancient duplication and fusion of a smaller ancestral protein related to the CMD and/or AhpD groups. Under this model, the N- and C-terminal regions of DehI should retain detectable evidence of common ancestry with one another and with the smaller proteins.

## The three protein groups

- **CMD proteins** are represented here by sequences assigned to Pfam family **PF02627**. The family is named for carboxymuconolactone decarboxylase and contains small, predominantly alpha-helical proteins with the compact structural unit relevant to this comparison. CMD proteins provide one model for a free-standing version of the smaller fold that may be related to a repeated unit within DehI. A Pfam assignment is used to define this collection because individual entries may have incomplete or inconsistent functional names.
- **AhpD proteins** are represented by the specific InterPro family **IPR004674**. AhpD is an antioxidant and redox protein best characterized for its role in transferring reducing equivalents within alkyl-hydroperoxide defence systems. Although its biochemical function differs from CMD and DehI, AhpD shares the broader structural theme of a small alpha-helical protein containing a related core. It therefore supplies a second group of free-standing proteins against which the proposed DehI repeats can be compared.
- **DehI-like proteins** are represented by InterPro family **IPR019714**, the family of configuration-inverting 2-haloacid dehalogenases also associated with Pfam **PF10778**. These enzymes remove halide from 2-haloacids and are substantially larger than the individual CMD or AhpD proteins. DehI is the proposed duplication-and-fusion product in this investigation: its two structural regions may preserve diverged copies of an ancestral unit resembling the smaller proteins. The experimentally determined DehI structure **PDB 3BJX** provides a reference for interpreting its architecture and locating a plausible internal boundary.

These labels define operational sequence groups rather than assuming that shared structure proves identical function or direct ancestry. In particular, the CMD Pfam entry describes a broader domain collection, whereas the AhpD and DehI InterPro identifiers are family-level assignments. That distinction will be kept in mind when interpreting similarities among the datasets.

## Reducing sequence redundancy by clustering

UniProt contains many nearly identical proteins, often because closely related species and strains have been sequenced repeatedly. If all of these records were included unchanged, densely sampled lineages could dominate an alignment or phylogenetic tree without adding much independent evolutionary information. To reduce this sampling redundancy, each of the three collections was clustered **independently** with MMseqs2. Independent clustering avoids allowing a sequence from one predefined group to replace a sequence from another group.

The primary datasets use a **90% amino-acid identity threshold**. Sequences connected at or above this threshold, with at least **80% bidirectional alignment coverage**, are grouped and one sequence is retained as the cluster representative. The coverage requirement prevents a short fragment from eliminating a substantially longer, full-length protein simply because the fragment is highly similar over the small region it covers. MMseqs2 also writes a membership table linking every original sequence to its representative, so the reduction is traceable and the discarded records are not conceptually lost.

The resulting `nr90` FASTA files are therefore non-redundant working sets rather than new biological families. Clustering at lower thresholds, such as 70% or 60%, produces progressively smaller and more diverse representative sets. Comparing conclusions across thresholds can help determine whether an inferred relationship is robust or is sensitive to the degree of redundancy reduction.

### Observed reduction at 70% identity

For the 70% identity working sets used below, clustering reduced the three collections as follows:

| Protein group | Original sequences | 70% representatives | Sequences removed | Retained |
|---|---:|---:|---:|---:|
| DehI-like | 127 | 78 | 49 | 61.4% |
| CMD-like | 68,845 | 18,084 | 50,761 | 26.3% |
| AhpD-like | 2,825 | 298 | 2,527 | 10.5% |

The AhpD collection underwent the strongest proportional reduction, retaining only about one sequence in ten, while the DehI collection retained more than half of its sequences. The much larger CMD collection remained large even after clustering, although almost three quarters of its entries were removed. These differences describe the redundancy and sequence-density of the downloaded datasets at this threshold; they should not, by themselves, be interpreted as evidence that one family is older, more diverse, or evolving faster than another. Database composition, taxonomic sampling, family-definition breadth, and the number of available proteomes can all affect the observed cluster counts.

The analysis develops this argument in stages. First, we inspect and quality-control the family-based sequence collections, using redundancy-reduced datasets so that densely sampled organisms do not dominate the comparisons. We then compare sequence lengths and domain organization to establish whether DehI has the architecture expected of a fusion product. Next, we examine the internal relationship between the two putative halves of DehI and compare each half independently with CMD and AhpD sequences. Finally, multiple-sequence alignments and phylogenetic reconstructions are used to ask whether the observed similarities are consistent with shared ancestry, duplication, and subsequent divergence.

Each step is intended to separate observation from interpretation. Structural resemblance or sequence similarity alone does not demonstrate a duplication-and-fusion event; support for the hypothesis should come from agreement among domain architecture, internal sequence symmetry, conserved residues, and phylogenetic placement. Alternative explanations—including convergence, remote homology without fusion, and biases introduced by sequence annotation or sampling—will therefore be considered alongside the proposed evolutionary model.
